```python
import os
import pandas as pd
import random
```

```python
IMAGE_PATH = os.path.join('pet_biometric_challenge_2022', 'train', 'images')
DATA_PATH = os.path.join('pet_biometric_challenge_2022', 'train')
CSV_PATH = os.path.join(DATA_PATH, 'train_data_sanitized.csv')
```

```python
df = pd.read_csv(CSV_PATH)
df.columns = ['dog_id', 'image']
df = df[df['dog_id'] < 500]
# len(df)
```

```python
random.seed(42)

positive_pairs = []
for dog_id in range(500):
    images = df[df['dog_id'] == dog_id]['image'].tolist()
    if len(images) >= 2:
        positive_pairs.append((images[0], images[1], 1))
```

```python
len(positive_pairs)
```

```python
negative_pairs = []
df = df[df['dog_id'] < 5650]
dog_ids = df['dog_id'].unique().tolist()

while len(negative_pairs) < 500:
    dog1, dog2 = random.sample(dog_ids, 2)  # ensure different dogs
    imgA = random.choice(df[df['dog_id'] == dog1]['image'].tolist())
    imgB = random.choice(df[df['dog_id'] == dog2]['image'].tolist())
    negative_pairs.append((imgA, imgB, 0))

pairs = positive_pairs + negative_pairs
random.shuffle(pairs)

pairs
```
